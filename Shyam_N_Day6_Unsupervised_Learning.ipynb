```python
import pandas as pd
import matplotlib.pyplot as plt
customers = pd.DataFrame({
    "AnnualSpend": [20,22,25,27,30,75,78,80,85,88,40,42],
    "VisitsPerMonth": [2,3,2,4,3,8,9,8,10,9,4,5]
})
plt.scatter(customers["AnnualSpend"],
            customers["VisitsPerMonth"])
plt.xlabel("Annual spend (hundreds)")
plt.ylabel("Visits per month")
plt.title("Customer data before clustering")
plt.show()

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

scaler = StandardScaler()
X_scaled = scaler.fit_transform(customers)

km = KMeans(n_clusters=3, random_state=42, n_init=10)
customers["Cluster"] = km.fit_predict(X_scaled)

plt.scatter(customers["AnnualSpend"],
            customers["VisitsPerMonth"],
            c=customers["Cluster"])
plt.xlabel("Annual spend")
plt.ylabel("Visits per month")
plt.title("K-Means customer segments")
plt.show()
print(customers)

from sklearn.cluster import DBSCAN

db = DBSCAN(eps=0.8, min_samples=2)
db_labels = db.fit_predict(X_scaled)

print("Cluster labels:", db_labels)
print("Noise points:", sum(db_labels == -1))

plt.scatter(customers["AnnualSpend"],
            customers["VisitsPerMonth"],
            c=db_labels)
plt.xlabel("Annual spend")
plt.ylabel("Visits per month")
plt.title("DBSCAN results (-1 means noise)")
plt.show()

from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

iris = load_iris()
X_iris = StandardScaler().fit_transform(iris.data)

pca = PCA(n_components=2)
X_2d = pca.fit_transform(X_iris)

print("Explained variance:",
      pca.explained_variance_ratio_.round(3))
plt.scatter(X_2d[:,0], X_2d[:,1], c=iris.target)
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("Iris data visualised using PCA")
plt.show()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import silhouette_score

data = {
    'Customer_ID': range(1, 21),
    'Annual_Spending': [
        15000, 18000, 20000, 22000, 25000,
        30000, 32000, 35000, 38000, 40000,
        45000, 48000, 50000, 52000, 55000,
        60000, 65000, 70000, 75000, 80000
    ],
    'Visits_Per_Month': [
        2, 3, 2, 4, 3,
        5, 6, 5, 7, 6,
        8, 9, 8, 10, 9,
        11, 12, 11, 13, 12
    ]
}

df = pd.DataFrame(data)


print("Customer Dataset:")
print(df)


X = df[['Annual_Spending', 'Visits_Per_Month']]


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)


kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)


print("K-Means Clustering Results:")
print(df)

# Visualise clusters
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Annual_Spending',
    y='Visits_Per_Month',
    hue='Cluster',
    palette='viridis',
    s=100
)

plt.title('Customer Segmentation using K-Means (K=3)')
plt.xlabel('Annual Spending')
plt.ylabel('Visits Per Month')
plt.legend(title='Cluster')
plt.grid(True)
plt.show()


average_spending = df.groupby('Cluster')['Annual_Spending'].mean()

print("Average Annual Spending for Each Cluster:")
print(average_spending)


k_values = [2, 3, 4]
inertia_values = []
silhouette_values = []

for k in k_values:
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)

    inertia_values.append(model.inertia_)
    silhouette_values.append(silhouette_score(X_scaled, labels))


results = pd.DataFrame({
    'K': k_values,
    'Inertia': inertia_values,
    'Silhouette Score': silhouette_values
})

print("K-Means Comparison Results:")
print(results)


plt.figure(figsize=(8, 5))
plt.plot(k_values, inertia_values, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 5))
plt.plot(k_values, silhouette_values, marker='o', color='green')
plt.title('Silhouette Score Comparison')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.grid(True)
plt.show()


dbscan = DBSCAN(eps=0.5, min_samples=3)
df['DBSCAN_Cluster'] = dbscan.fit_predict(X_scaled)


print("DBSCAN Clustering Results:")
print(df)


plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='Annual_Spending',
    y='Visits_Per_Month',
    hue='DBSCAN_Cluster',
    palette='rainbow',
    s=100
)

plt.title('Customer Segmentation using DBSCAN')
plt.xlabel('Annual Spending')
plt.ylabel('Visits Per Month')
plt.grid(True)
plt.show()


noise_points = df[df['DBSCAN_Cluster'] == -1]

print("Noise Points:")
print(noise_points)

```
